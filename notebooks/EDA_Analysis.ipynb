## 10 Key EDA Findings

### Finding 1 — SBI Mutual Fund dominates AUM
SBI Mutual Fund leads with Rs 12.5 lakh crore AUM (19.9% market share), nearly double ICICI Prudential at Rs 10.7L Cr. *(Chart 2)*

### Finding 2 — SIP industry grew 169% in 4 years
Monthly SIP inflows grew from Rs 11,517 Cr (Jan 2022) to an all-time high of Rs 31,002 Cr (Dec 2025), reflecting strong retail investor participation. *(Chart 3)*

### Finding 3 — Small Cap funds delivered the highest returns
SBI Small Cap Direct delivered 23.14% CAGR over 3 years — nearly double the Large Cap average of 12-13%, compensated by higher volatility. *(Chart 12)*

### Finding 4 — Folio count doubled — retail participation surging
Industry folios grew from 13.26 Cr (Jan 2022) to 26.12 Cr (Dec 2025), with Equity folios driving 75%+ of all growth. *(Chart 9)*

### Finding 5 — Banking and IT dominate equity portfolios
Banking (Rs 62,840 Cr, 30 funds) and IT (Rs 38,477 Cr, 27 funds) represent ~35% of all equity fund portfolios. *(Chart 11)*

### Finding 6 — 26-35 age group drives the most investment
The 26-35 Male cohort contributed Rs 64 Cr (18% of total volume) — the single largest investor demographic by both count and value. *(Chart 5)*

### Finding 7 — Large Cap funds are highly correlated
Pairwise return correlations among Large Cap funds range from 0.85-0.97, suggesting limited diversification benefit from holding multiple Large Cap funds. *(Chart 10)*

### Finding 8 — Liquid funds dominate category inflows
Liquid funds receive Rs 34,000-42,000 Cr net inflow every month — 5-8x higher than any other category — driven by institutional cash management. *(Chart 4)*

### Finding 9 — Punjab leads all states in investment volume
Punjab tops state-wise investment with Rs 315.8 Cr total, followed closely by Tamil Nadu (Rs 315.2 Cr). *(Chart 7)*

### Finding 10 — Direct plans significantly outperform Regular plans
Direct plan expense ratios are 0.6-1.0% lower than Regular plans, translating to 1-2% higher annual returns — compounding significantly over 5+ year horizons. *(Charts 12, 14)*
